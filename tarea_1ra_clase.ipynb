## Ejercicio 11 — El Salario de un Montaraz

```python
# Cálculo de pago base y bonos según escala de orcos derrotados
def calcular_salario_montaraz(orcos):
    pago_base = orcos * 35000
    if orcos > 100:
        bono_pct = 0.35
    elif orcos > 50:
        bono_pct = 0.20
    else:
        bono_pct = 0.0

    bono = pago_base * bono_pct
    total = pago_base + bono
    return pago_base, bono, total

base, bono, total = calcular_salario_montaraz(75)
print(f"Orcos: 75 -> Base: {base} monedas | Bono: {bono} monedas | Total: {total} monedas")
```

## Ejercicio 13 — El Verificador de Palantír

```python
# Uso del operador lógico XOR (^) para evaluar la conexión al Palantír
def probar_palantir(voluntad_fuerte, sangre_real):
    if voluntad_fuerte and sangre_real:
        return "Dominas el Palantír por completo."
    elif voluntad_fuerte ^ sangre_real:
        return "Logras usar el Palantír con gran esfuerzo."
    else:
        return "Sauron te corrompe."

print(f"V: True,  S: True  -> {probar_palantir(True, True)}")
print(f"V: True,  S: False -> {probar_palantir(True, False)}")
print(f"V: False, S: True  -> {probar_palantir(False, True)}")
print(f"V: False, S: False -> {probar_palantir(False, False)}")
```

## Ejercicio 21 — El Filtro de Gandalf

```python
# Separación de aliados y enemigos a partir de tuplas
def filtrar_aliados(personajes_lista):
    aliados_res = []
    enemigos_res = []
    for nombre_p, es_aliado in personajes_lista:
        if es_aliado:
            aliados_res.append(nombre_p)
        else:
            enemigos_res.append(nombre_p)
    return aliados_res, enemigos_res

personajes_prueba = [("Legolas", True), ("Saruman", False), ("Éowyn", True), ("Grima", False)]
aliados, enemigos = filtrar_aliados(personajes_prueba)

print("Aliados:", aliados)
print("Enemigos:", enemigos)
```

## Ejercicio 22 — Las Rimas del Anillo

```python
# Conteo de vocales totales y desglose por vocal
def contar_vocales(texto):
    conteo_total = 0
    conteo_individual = {'a': 0, 'e': 0, 'i': 0, 'o': 0, 'u': 0}

    for char in texto.lower():
        if char in "aeiou":
            conteo_total += 1
            conteo_individual[char] += 1

    return conteo_total, conteo_individual

frase = "Un anillo para gobernarlos a todos"
total_v, por_vocal = contar_vocales(frase)

print(f"Total de vocales en '{frase}': {total_v}")
print("Detalle por vocal:", por_vocal)
```

## Ejercicio 26 — Los Números de la Suerte de Gollum

```python
# Funciones matemáticas para análisis numérico
def es_par(n):
    return n % 2 == 0

def factorial(n):
    res = 1
    for i in range(1, n + 1):
        res *= i
    return res

def es_primo(n):
    if n < 2: return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0: return False
    return True

def fibonacci(n):
    if n <= 0: return []
    if n == 1: return [0]
    sec = [0, 1]
    while len(sec) < n:
        sec.append(sec[-1] + sec[-2])
    return sec

n_val = 10
print(f"¿{n_val} es par?:", es_par(n_val))
print(f"Factorial de {n_val}:", factorial(n_val))
print(f"¿{n_val} es primo?:", es_primo(n_val))
print(f"Primeros {n_val} términos de Fibonacci:", fibonacci(n_val))
```

## Ejercicio 27 — La Clase Personaje

```python
# Definición de la clase base Personaje
class Personaje:
    def __init__(self, nombre, raza, edad, vida=100):
        self.nombre = nombre
        self.raza = raza
        self.edad = edad
        self.vida = vida

    def presentarse(self):
        print(f"Soy {self.nombre}, un {self.raza} de {self.edad} años.")

    def recibir_dano(self, cantidad):
        self.vida = max(0, self.vida - cantidad)

    def esta_vivo(self):
        return self.vida > 0

p1 = Personaje("Aragorn", "Humano", 87)
p1.presentarse()
p1.recibir_dano(30)
print(f"Vida de Aragorn: {p1.vida} (¿Está vivo?: {p1.esta_vivo()})")
```

## Ejercicio 30 — Las Razas de la Tierra Media (Herencia I)

## Ejercicio 31 — El Mago Blanco (Herencia II)

```python
class Mago(Personaje):
    def __init__(self, nombre, edad, color="gris", mana=100):
        super().__init__(nombre, "Mago", edad)
        self.color = color
        self.mana = mana

    def lanzar_hechizo(self, nombre_hechizo, costo):
        if self.mana >= costo:
            self.mana -= costo
            print(f"🔮 {self.nombre} lanza {nombre_hechizo}. Maná restante: {self.mana}")
        else:
            print("⚡ Maná insuficiente.")

    def ascender(self):
        if self.color.lower() == "gris":
            self.color = "blanco"
            self.mana = 200
            print(f"✨ ¡{self.nombre} ha ascendido a Mago Blanco! Maná aumentado a 200.")

# Prueba
gandalf = Mago("Gandalf", 2000, "gris")
gandalf.lanzar_hechizo("Luz de Valinor", 40)
gandalf.ascender()
```

```python
class Personaje:
    def __init__(self, nombre, raza, edad, vida=100):
        self.nombre = nombre
        self.raza = raza
        self.edad = edad
        self.vida = vida

    def presentarse(self):
        print(f"Soy {self.nombre}, un {self.raza}.")

class Hobbit(Personaje):
    def __init__(self, nombre, edad):
        super().__init__(nombre, "Hobbit", edad)

    def presentarse(self):
        print(f"Soy {self.nombre} y me gustan las fiestas y el buen tabaco.")

class Elfo(Personaje):
    def __init__(self, nombre, edad):
        super().__init__(nombre, "Elfo", edad)

    def presentarse(self):
        print(f"Soy {self.nombre}, mi vista alcanza más allá del horizonte.")

class Enano(Personaje):
    def __init__(self, nombre, edad):
        super().__init__(nombre, "Enano", edad)

    def presentarse(self):
        print(f"¡Soy {self.nombre}, hijo de Glóin! Y no me llames pequeño.")

# Prueba polimórfica
comitiva = [
    Hobbit("Frodo", 33),
    Elfo("Legolas", 2931),
    Enano("Gimli", 139)
]

for p in comitiva:
    p.presentarse()
```

```python
class Inventario:
    def __init__(self, capacidad_maxima=5):
        self.capacidad_maxima = capacidad_maxima
        self.objetos = []

    def agregar_objeto(self, objeto):
        if len(self.objetos) < self.capacidad_maxima:
            self.objetos.append(objeto)
            print(f"📦 Se agregó '{objeto}' al inventario.")
        else:
            print(f"⚠️ El inventario está lleno. No se puede agregar '{objeto}'.")

    def MOSTRAR_INVENTARIO(self):
        print(f"--- Inventario ({len(self.objetos)}/{self.capacidad_maxima}) ---")
        for i, obj in enumerate(self.objetos, 1):
            print(f"{i}. {obj}")

# Prueba
mochila = Inventario(3)
mochila.agregar_objeto("Poción de vida")
mochila.agregar_objeto("Pan de Lembas")
mochila.agregar_objeto("Anillo Único")
mochila.agregar_objeto("Capa Elfica") # Supera el límite
mochila.MOSTRAR_INVENTARIO()
```

## Ejercicio 39 — El Sistema Completo de la Tierra Media

```python
import random

class Arma:
    def __init__(self, nombre, dano):
        self.nombre = nombre
        self.dano = dano

class PersonajeAventura:
    def __init__(self, nombre, raza, vida=100):
        self.nombre = nombre
        self.raza = raza
        self.vida = vida
        self.arma = None

    def equipar_arma(self, arma):
        self.arma = arma

# 1. Crear 6 personajes
p1, p2 = PersonajeAventura("Aragorn", "Humano"), PersonajeAventura("Legolas", "Elfo")
p3, p4 = PersonajeAventura("Gimli", "Enano"), PersonajeAventura("Frodo", "Hobbit")
p5, p6 = PersonajeAventura("Sam", "Hobbit"), PersonajeAventura("Gandalf", "Mago")

# 2. Asignar armas
p1.equipar_arma(Arma("Andúril", 30))
p2.equipar_arma(Arma("Arco Galadhrim", 25))

# 3. Agregar a Compañía
compania = [p1, p2, p3, p4, p5, p6]

# 4. Simular viaje de 5 etapas
print("🗺️ --- COMIENZA EL VIAJE DE LA COMPAÑÍA ---")
for etapa in range(1, 6):
    print(f"\n📍 Etapa {etapa}:")
    for p in compania:
        if p.vida > 0:
            dano = random.randint(5, 20)
            p.vida = max(0, p.vida - dano)
            print(f"- {p.nombre} recibe {dano} de daño. Vida: {p.vida}")

# 5. Sobrevivientes
print("\n🏆 --- ESTADO FINAL ---")
for p in compania:
    estado = f"Sobrevivió con {p.vida} de vida" if p.vida > 0 else "Cayó en batalla 💀"
    print(f"- {p.nombre} ({p.raza}): {estado}")
```

## Ejercicio 43 — El Inventario del Reino

```python
# Gestión de inventario mediante funciones
armeria = {"espadas": 340, "arcos": 120, "hachas": 87, "lanzas": 200}

def agregar_arma(nombre_item, cant):
    armeria[nombre_item] = armeria.get(nombre_item, 0) + cant

def usar_arma(nombre_item, cant):
    if nombre_item in armeria and armeria[nombre_item] >= cant:
        armeria[nombre_item] -= cant

def stock_total():
    return sum(armeria.values())

def arma_mas_abundante():
    return max(armeria, key=armeria.get)

agregar_arma("espadas", 50)
usar_arma("arcos", 20)
print("Stock Total:", stock_total())
print("Arma más abundante:", arma_mas_abundante())
```

## Ejercicio 44 — El Registro de Batallas por Reino

```python
# Agregación de datos numéricos en diccionarios
batallas = {
    "Gondor": [340, 120, 87],
    "Rohan": [500, 210],
    "Mordor": [1200, 890, 450, 300]
}

max_bajas = 0
reino_mas_bajas = ""

for reino, bajas_lista in batallas.items():
    tot = sum(bajas_lista)
    prom = tot / len(bajas_lista)
    print(f"Reino: {reino} | Total Bajas: {tot} | Promedio: {prom:.1f} | Batallas: {len(bajas_lista)}")
    if tot > max_bajas:
        max_bajas = tot
        reino_mas_bajas = reino

print(f"🏆 Reino con más bajas totales: {reino_mas_bajas} ({max_bajas})")
```

## Ejercicio 49 — El Parser del Pergamino Antiguo

```python
import re

# Transformación de texto delimitado a lista de diccionarios
pergamino = """nombre:Frodo|raza:Hobbit|edad:33|arma:Dardo
nombre:Legolas|raza:Elfo|edad:2931|arma:Arco
nombre:Gimli|raza:Enano|edad:139|arma:Hacha"""

lista_parsed = []
for linea in pergamino.strip().split("\n"):
    pares = re.findall(r"(\w+):([^|]+)", linea)
    dicc_p = {k: int(v) if v.isdigit() else v for k, v in pares}
    lista_parsed.append(dicc_p)

print(f"{'Nombre':<10} | {'Raza':<8} | {'Edad':<5} | {'Arma':<8}")
print("-" * 40)
for p in lista_parsed:
    print(f"{p['nombre']:<10} | {p['raza']:<8} | {p['edad']:<5} | {p['arma']:<8}")
```

## Ejercicio 50 — 🏆 El Concilio Final (Proyecto Integrador)

```python
import re
import random

# Sistema integrador final
class SistemaTierraMedia:
    def __init__(self):
        self.db_personajes = {}

    def registrar_personaje(self, personaje_obj):
        self.db_personajes[personaje_obj.nombre] = personaje_obj
        print(f"✅ Personaje '{personaje_obj.nombre}' registrado con éxito.")

    def listar_todos(self):
        print("\n=== LISTA COMPLETA ===")
        for p in self.db_personajes.values():
            print(p)

    def buscar_regex(self, patron_busqueda):
        print(f"\n--- Búsqueda patrón: '{patron_busqueda}' ---")
        for nombre_p, p in self.db_personajes.items():
            if re.search(patron_busqueda, nombre_p, re.IGNORECASE):
                print(p)

    def ver_estadisticas(self):
        total_p = len(self.db_personajes)
        prom_edad = sum(p.edad for p in self.db_personajes.values()) / total_p
        print("\n=== ESTADÍSTICAS DEL REINO ===")
        print(f"Total Personajes: {total_p} | Promedio Edad: {prom_edad:.1f} años")

sistema = SistemaTierraMedia()
sistema.registrar_personaje(Hobbit("Frodo Bolsón", 50))
sistema.registrar_personaje(Elfo("Legolas Hojaverde", 2931))
sistema.listar_todos()
sistema.buscar_regex("frodo")
sistema.ver_estadisticas()
```
